La COVID-19 (Genoma del SARS-CoV-2)Descripción del problemaUtilizar el genoma de referencia del SARS-CoV-2 con código de acceso NC_045512.   
Descargar el archivo en formato GenBank (gb).   
Recorrer la tabla de características (features) y filtrar las de tipo "gene".   
Extraer el nombre del gen desde los anotaciones (qualifiers.get("gene")).   
Mostrar un listado numerado con el nombre de cada gen y sus coordenadas de inicio y fin.   

```python
from Bio import Entrez
from Bio import SeqIO

Entrez.email = "lucia.cruz104@alu.ulpgc.es"

id_sars_cov2 = "NC_045512"

print("=== EJERCICIO 3: ANÁLISIS DE GENES EN SARS-CoV-2 ===")

# Descargar el genoma completo en formato GenBank
handle = Entrez.efetch(
    db="nucleotide",
    id=id_sars_cov2,
    rettype="gb",
    retmode="text"
)
registro_covid = SeqIO.read(handle, "genbank")
handle.close()

print(f"Registro descargado: {registro_covid.id}")
print(f"Organismo: {registro_covid.annotations.get('organism', 'SARS-CoV-2')}")
print(f"Longitud del genoma: {len(registro_covid.seq)} pb\n")

print("Coordenadas en base 1, con inicio y fin incluidos (estándar GenBank):")
print("-" * 60)
print(f"{'Nº':<4} {'Nombre del Gen':<18} {'Inicio (bp)':<15} {'Fin (bp)':<15}")
print("-" * 60)

contador_genes = 0
for feature in registro_covid.features:
    # Filtrar estrictamente anotaciones de tipo 'gene'
    if feature.type == "gene":
        contador_genes += 1
        nombre_gen = feature.qualifiers.get("gene", ["Sin nombre"])[0]
        
        # Transformación a coordenadas base 1 (inclusivas)
        inicio_base1 = int(feature.location.start) + 1
        fin_base1 = int(feature.location.end)
        
        print(f"{contador_genes:<4} {nombre_gen:<18} {inicio_base1:<15} {fin_base1:<15}")

print("-" * 60)
print(f"Total de genes anotados encontrados: {contador_genes}")
```
