La COVID-19 (Genoma del SARS-CoV-2)Descripción del problemaUtilizar el genoma de referencia del SARS-CoV-2 con código de acceso NC_045512.   
Descargar el archivo en formato GenBank (gb).   
Recorrer la tabla de características (features) y filtrar las de tipo "gene".   
Extraer el nombre del gen desde los anotaciones (qualifiers.get("gene")).   
Mostrar un listado numerado con el nombre de cada gen y sus coordenadas de inicio y fin.   

```python
from Bio import Entrez
from Bio import SeqIO

Entrez.email = "alumno.0000@ulpgc.es"

print("=== EJERCICIO 3: ANÁLISIS DE GENES EN EL GENOMA DEL SARS-CoV-2 (NC_045512) ===")

# 1. Descargar el genoma de referencia en formato GenBank
id_sars_cov2 = "NC_045512"
handle = Entrez.efetch(
    db="nucleotide",
    id=id_sars_cov2,
    rettype="gb",
    retmode="text"
)
registro_covid = SeqIO.read(handle, "genbank")
handle.close()

print(f"Organismo: {registro_covid.annotations.get('organism', 'SARS-CoV-2')}")
print(f"Longitud total del genoma: {len(registro_covid.seq)} nucleótidos\n")

# 2. Recorrer la lista de características y filtrar las de tipo "gene"
print("LISTADO NUMERADO DE GENES Y SUS COORDENADAS:")
print("-" * 55)
print(f"{'Nº':<4} {'Nombre del Gen':<18} {'Inicio':<10} {'Fin':<10}")
print("-" * 55)

contador_genes = 0
for feature in registro_covid.features:
    if feature.type == "gene":
        contador_genes += 1
        # Extraer el nombre del gen (o marcar como Desconocido si no existe la etiqueta)
        nombre_gen = feature.qualifiers.get("gene", ["Desconocido"])[0]
        
        # Coordenadas de inicio y fin (base 0 de Python)
        inicio = int(feature.location.start)
        fin = int(feature.location.end)
        
        print(f"{contador_genes:<4} {nombre_gen:<18} {inicio:<10} {fin:<10}")

print("-" * 55)
print(f"Total de genes identificados: {contador_genes}")
```
