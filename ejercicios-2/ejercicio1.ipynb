# Ejercicio 1: 
### Porcentaje GC y Traducción de SecuenciasDescripción del problemaTomar un fichero FASTA con varias secuencias de ADN y calcular el porcentaje de contenido GC ($G + C$) de cada una.   
### Traducir cada secuencia de ADN a su cadena de aminoácidos correspondiente.   
### Guardar las secuencias de aminoácidos en un nuevo archivo FASTA manteniendo los identificadores y descripciones originales.  

```python
import os
from Bio import SeqIO
from Bio.SeqRecord import SeqRecord

# Datos del trabajo
ASIGNATURA = "Bioinformática en Laboratorio"
INTEGRANTES = "Lucía Xufang Cruz Toste, Diego Muñoz Torres"
EMAIL_CONTACTO = "lucia.cruz102@alu.ulpgc.es"

fichero_entrada = "secuencias_dna.fasta"
fichero_salida = "proteinas_resultantes.fasta"

# Generación del archivo FASTA de prueba si no existe localmente
if not os.path.exists(fichero_entrada):
    secuencias_ejemplo = (
        ">seq1 Secuencia completa codificante\n"
        "ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG\n"
        ">seq2 Secuencia con nucleótidos incompletos\n"
        "ATGGCAGTTACCGA\n"
        ">seq3 Secuencia vacía de prueba\n"
        "\n"
    )
    with open(fichero_entrada, "w") as f:
        f.write(secuencias_ejemplo)

lista_proteinas = []

print(f"=== {ASIGNATURA} ===")
print(f"Autores: {INTEGRANTES}\n")
print("=== EJERCICIO 1: ANÁLISIS DE SECUENCIAS ADN Y TRADUCCIÓN ===")

for registro in SeqIO.parse(fichero_entrada, "fasta"):
    secuencia_adn = registro.seq
    longitud = len(secuencia_adn)
    
    print(f"\n--------------------------------------------------")
    print(f"ID: {registro.id}")
    print(f"Descripción: {registro.description}")
    print(f"Longitud ADN: {longitud} pb")
    
    # 1. Control de secuencias vacías
    if longitud == 0:
        print("ADVERTENCIA: La secuencia está vacía. No es posible calcular el porcentaje GC ni realizar la traducción.")
        continue
    
    # 2. Cálculo del Porcentaje GC
    conteo_g = secuencia_adn.count("G") + secuencia_adn.count("g")
    conteo_c = secuencia_adn.count("C") + secuencia_adn.count("c")
    porcentaje_gc = ((conteo_g + conteo_c) / longitud) * 100
    print(f"Porcentaje GC: {porcentaje_gc:.2f}%")
    
    # 3. Control de codón incompleto (múltiplo de 3)
    if longitud % 3 != 0:
        bases_sobrantes = longitud % 3
        print(f"ADVERTENCIA: La longitud ({longitud} pb) no es múltiplo de 3. "
              f"Quedan {bases_sobrantes} nucleótido(s) al final formando un codón incompleto.")
    
    # 4. Traducción a aminoácidos (Asumiendo sentido 5'->3' y marco de lectura correcto)
    secuencia_proteina = secuencia_adn.translate(to_stop=True)
    
    # 5. Mostrar la secuencia traducida por pantalla
    print(f"Secuencia de aminoácidos traducida: {secuencia_proteina}")
    
    # 6. Guardar el registro de la proteína conservando metadatos
    registro_prot = SeqRecord(
        secuencia_proteina,
        id=registro.id,
        description=registro.description
    )
    lista_proteinas.append(registro_prot)

# Exportar las proteínas traducidas
SeqIO.write(lista_proteinas, fichero_salida, "fasta")
print(f"\n¡Proceso completado! Se han exportado {len(lista_proteinas)} secuencias traducidas a '{fichero_salida}'.")
```
