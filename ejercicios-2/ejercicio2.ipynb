### Ejercicio 2: 
El Sarampión (Measles morbillivirus) Descripción del problema 
- Conectarse a NCBI Entrez e identificar el correo electrónico.   
- Utilizar Entrez.esearch en la base de datos nucleotide filtrando por la base de datos secundaria RefSeq (srcdb_refseq [Properties]) y la búsqueda "Measles morbillivirus [Organism]".   
- Extraer el primer ID devuelto (IdList[0]).   
- Descargar dicho registro en formato FASTA mediante Entrez.efetch e imprimir su descripción y los primeros 100 nucleótidos.   

```python
from Bio import Entrez
from Bio import SeqIO

Entrez.email = "lucia.cruz104@alu.ulpgc.es"

print("=== EJERCICIO 2: GENOMA DE REFERENCIA DEL VIRUS DEL SARAMPIÓN ===")

consulta = "Measles morbillivirus [Organism] AND srcdb_refseq [Properties]"

handle_busqueda = Entrez.esearch(db="nucleotide", term=consulta)
resultado_busqueda = Entrez.read(handle_busqueda)
handle_busqueda.close()

lista_ids = resultado_busqueda["IdList"]

if lista_ids:
    id_objetivo = lista_ids[0]
    print(f"ID RefSeq recuperado de forma automática: {id_objetivo}")

    handle_descarga = Entrez.efetch(
        db="nucleotide",
        id=id_objetivo,
        rettype="fasta",
        retmode="text"
    )
    registro_sarampion = SeqIO.read(handle_descarga, "fasta")
    handle_descarga.close()

    print("\n--- METADATOS DEL REGISTRO DESCARGADO ---")
    print(f"Acceso del registro: {registro_sarampion.id}")
    print(f"Longitud: {len(registro_sarampion.seq)} nucleótidos")
    print(f"Descripción completa: {registro_sarampion.description}")
    print(f"\nPrimeros 100 nucleótidos (5' -> 3'):\n{registro_sarampion.seq[:100]}")
    
    SeqIO.write(registro_sarampion, "sarampion_refseq.fasta", "fasta")
    print("\n[OK] Registro guardado en 'sarampion_refseq.fasta'.")
else:
    print("Error: No se encontraron registros en RefSeq para la búsqueda solicitada.")
```
