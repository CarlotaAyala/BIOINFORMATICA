Ejercicio 2: El Sarampión (Measles morbillivirus)Descripción del problemaConectarse a NCBI Entrez e identificar el correo electrónico.   Utilizar Entrez.esearch en la base de datos nucleotide filtrando por la base de datos secundaria RefSeq (srcdb_refseq [Properties]) y la búsqueda "Measles morbillivirus [Organism]".   Extraer el primer ID devuelto (IdList[0]).   Descargar dicho registro en formato FASTA mediante Entrez.efetch e imprimir su descripción y los primeros 100 nucleótidos.   

```python
from Bio import Entrez
from Bio import SeqIO

# Regla de oro NCBI: identificarse siempre con un correo electrónico
Entrez.email = "alumno.0000@ulpgc.es"

print("=== EJERCICIO 2: BÚSQUEDA Y DESCARGA DEL GENOMA DEL VIRUS DEL SARAMPIÓN ===")

# 1. Búsqueda en la base de datos de nucleótidos (filtrando por la base secundaria RefSeq)
consulta = "Measles morbillivirus [Organism] AND srcdb_refseq [Properties]"
handle_busqueda = Entrez.esearch(db="nucleotide", term=consulta)
resultado_busqueda = Entrez.read(handle_busqueda)
handle_busqueda.close()

# 2. Extraer el primer ID
lista_ids = resultado_busqueda["IdList"]
if lista_ids:
    id_objetivo = lista_ids[0]
    print(f"ID RefSeq encontrado automáticamente: {id_objetivo}")

    # 3. Descargar el registro en formato FASTA
    handle_descarga = Entrez.efetch(
        db="nucleotide",
        id=id_objetivo,
        rettype="fasta",
        retmode="text"
    )
    registro_sarampion = SeqIO.read(handle_descarga, "fasta")
    handle_descarga.close()

    # 4. Imprimir la descripción y los primeros 100 nucleótidos
    print("\n--- RESULTADOS DEL GENOMA ---")
    print(f"Descripción completa: {registro_sarampion.description}")
    print(f"Primeros 100 nucleótidos:\n{registro_sarampion.seq[:100]}")
    
    # Guardar en disco local opcionalmente
    SeqIO.write(registro_sarampion, "sarampion_refseq.fasta", "fasta")
    print("\nArchivo 'sarampion_refseq.fasta' guardado con éxito.")
else:
    print("No se encontraron resultados para la consulta.")
```
