Descripción y SelecciónDe la lista de proteínas propuesta, se han seleccionado tres proteínas fundamentales en biomedicina y fisiología:   
Hemoglobina (Cadena Beta - HBB): Proteína globular responsable del transporte de oxígeno desde los pulmones a los tejidos corporales.   Mioglobina (MB): Hemoproteína citoplasmática presente en las células musculares que almacena oxígeno para liberarlo durante el esfuerzo físico.   
Pepsina (PGA5 / Pepsina A): Enzima digestiva (proteasa) secretada en el estómago que degrada las proteínas de los alimentos en péptidos.

```python
from Bio import Entrez
from Bio import SeqIO

Entrez.email = "lucia.cruz104@alu.ulpgc.es"

proteinas_seleccionadas = [
    {
        "nombre_comun": "Subunidad Beta de Hemoglobina (HBB)",
        "id_refseq": "NP_000509.1",
        "gen": "HBB",
        "organismo": "Homo sapiens",
        "localizacion": "Citoplasma de eritrocitos",
        "funcion_detallada": "Componente estructural de la hemoglobina A. Se une al grupo hemo para transportar oxígeno desde los pulmones a los tejidos perféricos.",
        "url_ncbi": "https://www.ncbi.nlm.nih.gov/protein/NP_000509.1"
    },
    {
        "nombre_comun": "Mioglobina (MB)",
        "id_refseq": "NP_005359.1",
        "gen": "MB",
        "organismo": "Homo sapiens",
        "localizacion": "Sarcoplasma de miocitos (músculo esquelético y cardiaco)",
        "funcion_detallada": "Hemoproteína de unión a oxígeno que facilita la difusión celular de O2 y actúa como reserva de oxígeno muscular durante el ejercicio prolongado.",
        "url_ncbi": "https://www.ncbi.nlm.nih.gov/protein/NP_005359.1"
    },
    {
        "nombre_comun": "Precursor de Pepsina A-5 (PGA5 / Pepsinógeno)",
        "id_refseq": "NP_055039.1",
        "gen": "PGA5",
        "organismo": "Homo sapiens",
        "localizacion": "Luz estomacal / Secretada por células principales gástricas",
        "funcion_detallada": "Precursor inactivo (pepsinógeno) que se procesa proteolíticamente en medio ácido (pH < 2) produciendo la enzima activa pepsina A, encargada de la digestión de proteínas.",
        "url_ncbi": "https://www.ncbi.nlm.nih.gov/protein/NP_055039.1"
    }
]

print("=== EJERCICIO 5: DESCARGA Y FICHA TÉCNICA DE PROTEÍNAS CÉLEBRES ===")

for info in proteinas_seleccionadas:
    # Descarga del registro FASTA desde la base de datos 'protein'
    handle = Entrez.efetch(
        db="protein",
        id=info["id_refseq"],
        rettype="fasta",
        retmode="text"
    )
    registro = SeqIO.read(handle, "fasta")
    handle.close()
    
    # Guardar archivo FASTA localmente
    nombre_fasta = f"{info['id_refseq']}.fasta"
    SeqIO.write(registro, nombre_fasta, "fasta")
    
    print("\n" + "=" * 70)
    print(f"PROTEÍNA: {info['nombre_comun']}")
    print("=" * 70)
    print(f"  • Identificador RefSeq: {info['id_refseq']}")
    print(f"  • Descripción del registro: {registro.description}")
    print(f"  • Organismo: {info['organismo']}")
    print(f"  • Gen asociado: {info['gen']}")
    print(f"  • Localización subcelular/tisular: {info['localizacion']}")
    print(f"  • Función biológica: {info['funcion_detallada']}")
    print(f"  • Longitud: {len(registro.seq)} aminoácidos")
    print(f"  • Secuencia (Vista previa primeros 40 aa): {registro.seq[:40]}...")
    print(f"  • Enlace NCBI: {info['url_ncbi']}")
    print(f"  • Archivo guardado: '{nombre_fasta}'")

print("\n" + "=" * 70)
print("¡Descarga de los 3 archivos FASTA completada con éxito!")
```
