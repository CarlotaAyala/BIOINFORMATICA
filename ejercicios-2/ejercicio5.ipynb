Ejercicio 5: Algunas proteínas "célebres"
Descripción y SelecciónDe la lista de proteínas propuesta, se han seleccionado tres proteínas fundamentales en biomedicina y fisiología:   
Hemoglobina (Cadena Beta - HBB): Proteína globular responsable del transporte de oxígeno desde los pulmones a los tejidos corporales.   Mioglobina (MB): Hemoproteína citoplasmática presente en las células musculares que almacena oxígeno para liberarlo durante el esfuerzo físico.   
Pepsina (PGA5 / Pepsina A): Enzima digestiva (proteasa) secretada en el estómago que degrada las proteínas de los alimentos en péptidos.

```python
from Bio import Entrez
from Bio import SeqIO

Entrez.email = "alumno.0000@ulpgc.es"

# Selección de 3 proteínas célebres con sus IDs de referencia NCBI/RefSeq en la base de datos 'protein'
proteinas_celebres = [
    {
        "nombre": "Hemoglobina (Cadena Beta - HBB)",
        "id_protein": "NP_000509.1",
        "descripcion_biologica": "Transportador primario de oxígeno en los eritrocitos de la sangre."
    },
    {
        "nombre": "Mioglobina (MB)",
        "id_protein": "NP_002326.1",
        "descripcion_biologica": "Almacena oxígeno en el tejido muscular para su liberación durante el ejercicio."
    },
    {
        "nombre": "Pepsina A (PGA5)",
        "id_protein": "NP_000918.1",
        "descripcion_biologica": "Enzima digestiva gástrica encargada de la hidrólisis de enlaces peptídicos."
    }
]

print("=== EJERCICIO 5: DESCARGA Y ANÁLISIS DE PROTEÍNAS CÉLEBRES ===")

for info in proteinas_celebres:
    # Descargar el registro FASTA desde la base de datos 'protein'
    handle = Entrez.efetch(
        db="protein",
        id=info["id_protein"],
        rettype="fasta",
        retmode="text"
    )
    registro = SeqIO.read(handle, "fasta")
    handle.close()
    
    # Guardar en archivo .fasta local
    nombre_archivo = f"{info['id_protein']}.fasta"
    SeqIO.write(registro, nombre_archivo, "fasta")
    
    # Imprimir resumen de información
    print(f"\n Proteína: {info['nombre']}")
    print(f"  • Función biológica: {info['descripcion_biologica']}")
    print(f"  • ID RefSeq Proteína: {registro.id}")
    print(f"  • Longitud: {len(registro.seq)} aminoácidos")
    print(f"  • Secuencia (primeros 50 aa): {registro.seq[:50]}...")
    print(f"  • Archivo guardado: '{nombre_archivo}'")
```
