Hemoglobina (Cadena Beta - HBB): Proteína globular responsable del transporte de oxígeno desde los pulmones a los tejidos corporales.   Mioglobina (MB): Hemoproteína citoplasmática presente en las células musculares que almacena oxígeno para liberarlo durante el esfuerzo físico.

```python
from Bio import Entrez
from Bio import SeqIO

Entrez.email = "lucia.cruz104@alu.ulpgc.es"

proteinas_seleccionadas = [
    {
        "nombre_comun": "Colágeno tipo I, cadena alfa 1 (precursor)",
        "id_refseq": "NP_000079.2",
        "gen": "COL1A1",
        "organismo": "Homo sapiens",
        "localizacion": "Matriz extracelular de tejidos como piel, huesos y tendones",
        "funcion_detallada": (
            "Precursor de la cadena alfa 1 del colágeno tipo I, "
            "una proteína estructural que proporciona resistencia "
            "mecánica a los tejidos."
        ),
        "url_ncbi": "https://www.ncbi.nlm.nih.gov/protein/NP_000079.2"
    },
    {
        "nombre_comun": "Glucagón (precursor: preproglucagón)",
        "id_refseq": "NP_002045.1",
        "gen": "GCG",
        "organismo": "Homo sapiens",
        "localizacion": "El glucagón se secreta por las células alfa del páncreas",
        "funcion_detallada": (
            "Precursor que se procesa para originar varios péptidos, "
            "entre ellos el glucagón. Esta hormona favorece el aumento "
            "de la glucosa en sangre mediante su acción sobre el hígado."
        ),
        "url_ncbi": "https://www.ncbi.nlm.nih.gov/protein/NP_002045.1"
    },
    {
        "nombre_comun": "Subunidad beta de hemoglobina (HBB)",
        "id_refseq": "NP_000509.1",
        "gen": "HBB",
        "organismo": "Homo sapiens",
        "localizacion": "Citoplasma de los eritrocitos",
        "funcion_detallada": (
            "Forma parte de la hemoglobina A y se asocia a un grupo "
            "hemo que permite la unión al oxígeno, contribuyendo "
            "a su transporte desde los pulmones hasta los tejidos."
        ),
        "url_ncbi": "https://www.ncbi.nlm.nih.gov/protein/NP_000509.1"
    }
]

print("=== EJERCICIO 5: DESCARGA Y FICHA TÉCNICA DE PROTEÍNAS CÉLEBRES ===")

for info in proteinas_seleccionadas:
    handle = Entrez.efetch(
        db="protein",
        id=info["id_refseq"],
        rettype="fasta",
        retmode="text"
    )
    registro = SeqIO.read(handle, "fasta")
    handle.close()
    
    nombre_fasta = f"{info['id_refseq']}.fasta"
    SeqIO.write(registro, nombre_fasta, "fasta")
    
    print("\n" + "=" * 70)
    print(f"PROTEÍNA: {info['nombre_comun']}")
    print("=" * 70)
    print(f"  • Identificador RefSeq: {info['id_refseq']}")
    print(f"  • Descripción del registro: {registro.description}")
    print(f"  • Organismo: {info['organismo']}")
    print(f"  • Gen asociado: {info['gen']}")
    print(f"  • Localización subcelular/tisular: {info['localizacion']}")
    print(f"  • Función biológica: {info['funcion_detallada']}")
    print(f"  • Longitud: {len(registro.seq)} aminoácidos")
    print(f"  • Secuencia (Vista previa primeros 40 aa): {registro.seq[:40]}...")
    print(f"  • Enlace NCBI: {info['url_ncbi']}")
    print(f"  • Archivo guardado: '{nombre_fasta}'")

print("\n" + "=" * 70)
print("¡Descarga de los 3 archivos FASTA completada con éxito!")
```
